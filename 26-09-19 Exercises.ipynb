Problem 3
Write code to calculate the sum of all numbers from 1 up to a given number n.

```python
n = 4
i = 1
s = 0

while i <= n:
    s = s + i
    i = i + 1

print (s)
```

Problem 4
Write code to print the formatted multiplication table for a given number n.

n=1

1x1=1
1x2=2
1x3=3

```python
m = 1

while m <= 10:
    n = 1
    
    while n <= 10:
    
        print (n, '*', m, '=', n*m)
        n = n + 1

    print(" ")
    m = m + 1
```
